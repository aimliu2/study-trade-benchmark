# P1 baseline — 1D OHLC, 2026 Q1–Q3

Plain daily candles for the P1 universe. Bars come from the replayer engine's validated loader (via `helpers/replayer_bars.py`).
The x axis is one slot per bar, so weekends and holidays leave no gaps; dates are bar open (UTC).

- **XAUT** has no XAUTUSDT bars in the replayer data yet — its chart uses MT5 **XAUUSD** spot gold as a stand-in.

```python
from helpers.replayer_bars import DATA_SYMBOL, period_bars, plot_ohlc

SYMBOLS = ["BTC", "ETH", "XAUT", "QQQ"]
TIMEFRAME, START, END = "1d", "2026Q1", "2026Q3"
LABEL = "2026 Q1–Q3"

bars = {symbol: period_bars(symbol, TIMEFRAME, START, END) for symbol in SYMBOLS}
for symbol, price in bars.items():
    print(f"{symbol:<5} ({DATA_SYMBOL[symbol]}): {len(price)} daily bars, {price.index[0]:%Y-%m-%d} -> {price.index[-1]:%Y-%m-%d}")
```

```python
plot_ohlc(bars["BTC"], f"BTCUSDT · {LABEL} · plain 1D", "BTCUSDT")
```

```python
plot_ohlc(bars["ETH"], f"ETHUSDT · {LABEL} · plain 1D", "ETHUSDT")
```

```python
plot_ohlc(bars["XAUT"], f"XAUT (proxy: MT5 XAUUSD) · {LABEL} · plain 1D", "XAUUSD")
```

```python
plot_ohlc(bars["QQQ"], f"QQQ · {LABEL} · plain 1D", "QQQ")
```
